```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/content/placement123.csv')
df.sample(10)
```

```python
df['cgpa'].skew()
```

```python
df['placement_exam_marks'].skew()
```

```python
upper_limit = df['cgpa'].mean() + 3*df['cgpa'].std()
lower_limit = df['cgpa'].mean() - 3*df['cgpa'].std()
print(upper_limit)
print(lower_limit)
```

```python
df[(df['cgpa']>upper_limit) | (df['cgpa']<lower_limit)]
```

TRIMMING

```python
new_df = df[(df['cgpa']<upper_limit) & (df['cgpa']>lower_limit)]
new_df
```

CAPPING

```python
df['cgpa'] = np.where(
    df['cgpa']>upper_limit,
    upper_limit,
    np.where(df['cgpa']<lower_limit,
    lower_limit,
    df['cgpa']
)
)
```
