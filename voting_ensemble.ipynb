```python
import pandas as pd
import numpy as np
df = pd.read_csv('/content/haberman.csv')
```

```python
df.head(5)
```

```python
df['age'] = df.iloc[:,0]
df['year'] = df.iloc[:,1]
df['survived'] = df.iloc[:,3]
```

```python
df = df.drop('30',axis=1)
df = df.drop('64',axis=1)
df = df.drop('1',axis=1)
df = df.drop('1.1',axis=1)

```

```python
df.head(10)
```

```python
x = df.drop('survived',axis=1)
y = df['survived']
```

```python
from sklearn.model_selection import cross_val_score
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
lr = LogisticRegression()
svm = SVC(probability=True)
dt = DecisionTreeClassifier()
print(np.mean(cross_val_score(lr,x,y,cv=10,scoring='accuracy')))
print(np.mean(cross_val_score(dt,x,y,cv=10,scoring='accuracy')))
print(np.mean(cross_val_score(svm,x,y,cv=10,scoring='accuracy')))
```

```python
from sklearn.ensemble import VotingClassifier
vc = VotingClassifier(estimators = [('lr',lr),('svm',svm),('dt',dt)],voting='hard')
print(np.mean(cross_val_score(vc,x,y,cv=10,scoring='accuracy')))
```

```python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vc = VotingClassifier(estimators = [('lr',lr),('svm',svm),('dt',dt)],voting='hard',weights=[i,j,k])
      print("for i = ",i,"for j =",j,"for k =",k,np.mean(cross_val_score(vc,x,y,cv=10,scoring='accuracy')))
```
