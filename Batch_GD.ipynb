```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=100,n_features=5,n_informative=5,n_targets=1,noise=54)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
print(x_train.shape)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
print(lr.coef_)
print(lr.intercept_)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,lr.predict(x_test))
```

```python
class GDregressor:

  def __init__(self,lr = 0.01,epochs =100):
    self.coef_ = None
    self.intercept_ = None
    self.lr = lr
    self.epochs = epochs

  def fit(self,x_train,y_train):
    self.intercept_ = 0
    self.coef_ = np.ones(x_train.shape[1])

    for i in range(self.epochs):
      y_hat = self.intercept_ + np.dot(x_train,self.coef_)
      der_int = -2*np.mean(y_train - y_hat)
      self.intercept_ = self.intercept_ - (self.lr*der_int)
      der_coe = -2*(np.dot(y_train - y_hat,x_train))/(x_train.shape[0])
      self.coef_ = self.coef_ - (self.lr*der_coe)

    print(self.coef_)
    print(self.intercept_)

  def predict(self,x_test):
    return self.intercept_ + np.dot(x_test,self.coef_)
```

```python
gd = GDregressor(lr=0.01,epochs=6000)
gd.fit(x_train,y_train)
```

```python
y_pred = gd.predict(x_test)
r2_score(y_test,y_pred)
```
