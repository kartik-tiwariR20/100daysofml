```python
from sklearn.datasets import make_regression
import numpy as np
```

```python
x,y = make_regression(n_samples=10,n_features=1,n_informative=1,n_targets=1,noise=4,random_state=13)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x,y)
print(lr.coef_)
print(lr.intercept_)      #hum log isi ko ie b ko nikaalenge apni class se
```

```python
class GDRegressor:
  def __init__(self,lr,epochs):
    self.m = 26.77311963
    self.b = 120
    self.lr = lr
    self.epoch = epochs


  def fit(self,x,y):
    for i in range(self.epoch):
      loss_slope = -2*np.sum(y - self.m*x.ravel() - self.b)
      step_size = self.lr*loss_slope
      self.b = self.b - step_size
      print(self.b," ",loss_slope)
    print(self.b)
```

```python
gd = GDRegressor(0.01,100)
gd.fit(x,y)
```

```python
class gradregressor:
  def __init__(self,lr,epochs):
    self.m = 100
    self.b = 120
    self.lr = lr
    self.epoch = epochs


  def fit(self,x,y):
    for i in range(self.epoch):
      loss_slope_b = -2*np.sum(y - self.m*x.ravel() - self.b)
      step_size = self.lr*loss_slope_b
      self.b = self.b - step_size
      loss_slope_m = -2*np.sum((y - self.m*x.ravel() - self.b)*x.ravel())
      self.m = self.m - (self.lr*loss_slope_m)
    print(self.b,self.m)

  def predict(self,x):
    return self.m*x + self.b
```

```python
reg = gradregressor(0.05,40)
reg.fit(x,y)
```

```python
for x in range(10):
  print(reg.predict(x))
```
