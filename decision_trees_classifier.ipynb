```python
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris # Needed to get feature_names
import pandas as pd # Needed for DataFrame

# Re-load iris dataset to get the feature names and construct the DataFrame
iris = load_iris()

# Create a DataFrame from x (features) with feature names
df = pd.DataFrame(x, columns=iris.feature_names)

# Add y (target) as a new column
df['target'] = y

plt.figure(figsize=(8, 6))
sns.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='target', data=df, palette='viridis')
plt.title('Iris Dataset: Sepal Length vs. Sepal Width (colored by target)')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.legend(title='Target Class')
plt.grid(True)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
```

```python
def best(criterion,x,splitter,maxleaf,y):
  from sklearn.tree import DecisionTreeClassifier, plot_tree
  import matplotlib.pyplot as plt

  dt = DecisionTreeClassifier(criterion=criterion,max_depth=x,splitter=splitter,max_leaf_nodes=maxleaf,min_samples_split=y)
  dt.fit(x_train,y_train)
  from sklearn.metrics import accuracy_score
  r = accuracy_score(y_test,dt.predict(x_test))
  print(r)

  plt.figure(figsize=(20,10))
  plot_tree(dt, feature_names=iris.feature_names, class_names=iris.target_names.tolist(), filled=True, rounded=True)
  plt.title(f"Decision Tree with Criterion: {criterion}, Max Depth: {x}, Splitter: {splitter}, Max Leaf Nodes: {maxleaf}, Min Samples Split: {y}")
  plt.show()
```

```python
best("gini",2,"random",3,10)
```

```python
best("gini",10,"best",8,10)
```
