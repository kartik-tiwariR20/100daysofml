```python
from sklearn.datasets import make_regression
x,y = make_regression(n_samples=1000,n_features=5,n_informative=4,n_targets=1,noise=127,random_state=42)
```

```python
import matplotlib.pyplot as plt
```

```python
from sklearn.model_selection import train_test_split,cross_val_score
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.ensemble import AdaBoostRegressor
adr = AdaBoostRegressor(n_estimators=100)
adr.fit(x_train,y_train)
from sklearn.metrics import r2_score
r2_score(y_test,adr.predict(x_test))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gbr = GradientBoostingRegressor()
gbr.fit(x_train,y_train)
r2_score(y_test,gbr.predict(x_test))
```

```python
from xgboost import XGBRegressor
xg = XGBRegressor(n_estimators=200, learning_rate=0.05, max_depth=4, random_state=42)
xg.fit(x_train,y_train)
r2_score(y_test,xg.predict(x_test))
```
