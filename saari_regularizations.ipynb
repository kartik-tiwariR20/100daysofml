```python
from sklearn.datasets import make_regression
x,y = make_regression(n_samples=500,n_features=50,n_informative=45,n_targets=1,noise=87)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2)
x_train
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(x_train,y_train)
from sklearn.metrics import r2_score
r2_score(y_test,lr.predict(x_test))
```

```python
from sklearn.linear_model import Ridge
rid = Ridge(alpha=0.5)
rid.fit(x_train,y_train)
from sklearn.metrics import r2_score
r2_score(y_test,rid.predict(x_test))
```

```python
from sklearn.linear_model import Lasso
ls = Lasso(alpha=0.5)
ls.fit(x_train,y_train)
from sklearn.metrics import r2_score
r2_score(y_test,ls.predict(x_test))
```

```python
from sklearn.linear_model import ElasticNet
en = ElasticNet(alpha=0.0005,l1_ratio=0.9)
en.fit(x_train,y_train)
from sklearn.metrics import r2_score
r2_score(y_test,en.predict(x_test))
```
