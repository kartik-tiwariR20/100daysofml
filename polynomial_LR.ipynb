```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Generate a random number of entries between 100 and 200
n_entries = np.random.randint(100, 201)

# Generate x values
X = np.linspace(-5, 5, n_entries)

# Define a polynomial function (e.g., quadratic) with some coefficients
# y = 0.5 * x^2 + 2 * x + 1
y_true = 0.5 * X**2 + 2 * X + 1

# Add random noise to y values
noise = np.random.normal(0, 2, n_entries) # Mean 0, standard deviation 2
y = y_true + noise

# Create a DataFrame
dataset = pd.DataFrame({'x': X, 'y': y})

print(f"Created a dataset with {n_entries} entries.")
print("First 5 rows of the dataset:")
print(dataset.head())

# Visualize the dataset
plt.figure(figsize=(10, 6))
plt.scatter(dataset['x'], dataset['y'], label='Generated Data', alpha=0.7)
plt.plot(X, y_true, color='red', linestyle='--', label='True Polynomial Function')
plt.title('Synthetic Dataset for Polynomial Regression')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(X,y,test_size=0.2)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
lr = LinearRegression()
# Reshape x_train and x_test to be 2D arrays
lr.fit(x_train.reshape(-1, 1),y_train)

r2_score(y_test,lr.predict(x_test.reshape(-1, 1)))
```

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
x_train_trf = poly.fit_transform(x_train.reshape(-1, 1))
x_test_trf = poly.transform(x_test.reshape(-1, 1))
```

```python
lr.fit(x_train_trf,y_train)

r2_score(y_test,lr.predict(x_test_trf))
```
