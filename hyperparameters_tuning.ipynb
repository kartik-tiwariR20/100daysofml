```python
import pandas as pd
import numpy as np
df =pd.read_csv('/content/heart.csv')
df.head(10)
```

```python
df.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df.drop('target',axis=1),df['target'],test_size=0.2,random_state=42)
```

```python
from math import log
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import cross_val_score

lr = LogisticRegression()
svm = SVC(probability=True,kernel='linear')
dtc = DecisionTreeClassifier(criterion = 'gini')
lr.fit(x_train,y_train)
dtc.fit(x_train,y_train)
svm.fit(x_train,y_train)
print(np.mean(cross_val_score(lr,x_train,y_train,cv=5,scoring="accuracy")))
print(np.mean(cross_val_score(svm,x_train,y_train,cv=5,scoring="accuracy")))
print(np.mean(cross_val_score(dtc,x_train,y_train,cv=5,scoring="accuracy")))
```

```python
from sklearn.ensemble import BaggingClassifier
from sklearn.ensemble import RandomForestClassifier
bc = BaggingClassifier(estimator=dtc,n_estimators=10)
bc.fit(x_train,y_train)
from sklearn.metrics import accuracy_score
print(accuracy_score(y_test,bc.predict(x_test)))

rfc = RandomForestClassifier(n_estimators=10)
rfc.fit(x_train,y_train)
print(accuracy_score(y_test,rfc.predict(x_test)))
```

```python
params = {
    'n_estimators' : [5,10,15,20],
    'criterion' : ['gini','entropy']
}
```

```python
from sklearn.model_selection import GridSearchCV
gsc = GridSearchCV(estimator = rfc , param_grid=params)
gsc.fit(x_train,y_train)
gsc.best_params_
```
