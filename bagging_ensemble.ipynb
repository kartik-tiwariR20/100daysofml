```python
from sklearn.datasets import make_classification
x, y = make_classification(n_samples=3000, n_features=3, n_informative=3, n_redundant=0, n_classes=2, class_sep=0.5, random_state=42)
```

```python
from numpy.random import rand
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
```

```python
lr = LogisticRegression()
svm = SVC(probability=True,kernel='linear')
dtc = DecisionTreeClassifier(criterion="gini",max_depth=4)
lr.fit(x_train,y_train)
svm.fit(x_train,y_train)
dtc.fit(x_train,y_train)
```

```python
print(accuracy_score(y_test,lr.predict(x_test)))
print(accuracy_score(y_test,svm.predict(x_test)))
print(accuracy_score(y_test,dtc.predict(x_test)))
```

```python
from sklearn.ensemble import BaggingClassifier
import numpy as np
bc = BaggingClassifier(estimator = lr , n_estimators=100,max_samples = 0.50,bootstrap=True,bootstrap_features=False)
bc.fit(x_train,y_train)
np.mean(cross_val_score(bc,x,y,cv=10,scoring="accuracy"))
```

```python
bc2 = BaggingClassifier(estimator = svm , n_estimators=100,max_samples = 0.50,bootstrap=True,bootstrap_features=False)
bc2.fit(x_train,y_train)
np.mean(cross_val_score(bc2,x,y,cv=10,scoring="accuracy"))
```

```python
bc3 = BaggingClassifier(estimator = dtc , n_estimators=100,max_samples = 0.50,bootstrap=True,bootstrap_features=False)
bc3.fit(x_train,y_train)
np.mean(cross_val_score(bc3,x,y,cv=10,scoring="accuracy"))
```
