```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import make_regression
x,y = make_regression(n_samples=300,n_features=2,n_informative=2,n_targets=1,noise=44,random_state=42)
x
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:,0],x[:,1],c=y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.preprocessing import PolynomialFeatures
pf = PolynomialFeatures(degree=2)
x_train_trf = pf.fit_transform(x_train)
x_test_trf = pf.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
lr = LinearRegression()
dtr = DecisionTreeRegressor()
lr.fit(x_train_trf,y_train)
dtr.fit(x_train,y_train)
from sklearn.metrics import r2_score
print(r2_score(y_test,lr.predict(x_test_trf)))
print(r2_score(y_test,dtr.predict(x_test)))
```

```python
from sklearn.ensemble import RandomForestRegressor
rfr = RandomForestRegressor(n_estimators = 100)
rfr.fit(x_train,y_train)
print(r2_score(y_test,rfr.predict(x_test)))
```

```python
from sklearn.ensemble import GradientBoostingRegressor
gbr = GradientBoostingRegressor(n_estimators=100)
gbr.fit(x_train,y_train)
print(r2_score(y_test,gbr.predict(x_test)))
```
